# Étape 2 — Extraction des champs clés

Ce notebook importe maintenant la logique d'extraction depuis `src/field_extraction.py`
(factorisée depuis la v2 -- voir `docs/JOURNAL_V2.md`) au lieu de la recopier.
Il ne garde que la configuration et un test de démonstration sur un dossier réel.

```python
import sys
from pathlib import Path
sys.path.append(str(Path("..").resolve()))

import json
from pathlib import Path
from src.field_extraction import extraire_champs_page, LABELS

OCR_RESULTS_DIR = Path("../outputs/ocr_results")


def charger_ocr(dossier_json_path: Path) -> dict:
    with open(dossier_json_path, "r", encoding="utf-8") as f:
        return json.load(f)
```

## Test sur un dossier réel (étape 1 déjà exécutée)

```python
exemple = charger_ocr(OCR_RESULTS_DIR / "Dossier_000001_ocr.json")

for page in exemple["pages"]:
    print(f"--- Page {page['page']} ---")
    champs = extraire_champs_page(page)
    for nom_champ, resultat in champs.items():
        conf = f"{resultat['confiance']:.2f}" if resultat["confiance"] is not None else "N/A"
        print(f"  {nom_champ}: {resultat['valeur']!r} (confiance={conf})")
    print()
```
